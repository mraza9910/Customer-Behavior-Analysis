# Problem Statement:

A leading retail company wants to better understand its customers shopping behavior in order to improve sales, customer satisfaction, and long term loyalty.  The management team has noticed changes in purchasing patterns across demographics, product categories, and sales channels (online v. offline).  They are particularly interested in uncovering which factors, such as discounts, reviews, seasons, or payment preferences, drive customer decisions and repeat purchases.

## Business Question We Are Answering:

How can the company leverage customer shopping data to identify trends, improve customer engagement, and optimize marketing and product strategies? 

```python
import pandas as pd

df = pd.read_csv('/Users/maidaraza/Desktop/Data Analytics Projects/2026-10-05 - Customer Sales Behaviour/Dataset/customer_shopping_behavior.csv')
```

```python
# There are 4 product categories.  Hence, we cannot simply replace the missing values with the median ratings.  
# Instead, we will calculate the median rating for each product and replace the missings with that

df["Review Rating"] = df.groupby("Category")['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum() # No nulls left anymore!
```

```python
# Putting the column names in snake_case. This makes the code consistent and easier to read

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.columns # all the columns have been converted to snake case
```

```python
df = df.rename(columns={"purchase_amount_(usd)": "purchase_amount"})
```

```python
df.columns
```

```python
# Feature Engineering - we will create customer groups here

labels = ["Young Adult", "Adult", "Middle-Aged", "Senior"]
df["age_group"] = pd.qcut(df['age'], q=4, labels = labels) # splits the ages into 4 equal sized groups
```

```python
df.columns
```

```python
df[["age", "age_group"]].head()
```

```python
# We will create another feature called the purchase frequency days.  
# Right now, we have the textual data on purchase frequency; we will convert that into numbers here
# We create a dictionary to be able to map the days

frequency_mapping = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Monthly": 30,
    "Quarterly": 90,
    "Bi-Weekly": 14,
    "Annually": 365,
    "Every 3 Months": 90
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["purchase_frequency_days", "frequency_of_purchases"]].head()
```

```python
# Here we are exploring if discount applied and promo code used are essentially the same columns

df[["discount_applied", "promo_code_used"]].head()
```

```python
# This tells us that these two columns are the same and one is redundant

(df["discount_applied"] == df["promo_code_used"]).all()
```

```python
df = df.drop("promo_code_used", axis = 1) # axis = 1 tells to drop the column
```
